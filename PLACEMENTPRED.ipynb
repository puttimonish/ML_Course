```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset with the correct name
df = pd.read_csv("placement_predict_50k Dataset (2).csv")

# Dataset Overview
print(df.head())        # first 5 rows
print(df.tail())        # last 5 rows
print(df.shape)         # rows, columns
print(df.info())        # column types + nulls
print(df.describe())    # summary stats
print(df.columns.tolist())  # list of column names
print(df.sample(5))     # random 5 rows
print(df.nunique())     # unique values per column
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="PlacementStatus")
plt.title("Placement Status Distribution")
plt.show()

plt.figure(figsize=(6,4))
sns.kdeplot(df["CodingTestScore"], fill=True)
plt.title("Coding Test Score Density")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x=df["CGPA"])
plt.title("CGPA Box Plot")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="CGPA", y="AttendancePercent")
plt.title("CGPA vs Attendance Percentage")
plt.show()

plt.figure(figsize=(6,4))
sns.regplot(data=df, x="CGPA", y="CodingTestScore")
plt.title("CGPA vs Coding Test Score")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(data=df, x="PlacementStatus", y="CGPA")
plt.title("CGPA vs Placement Status")
plt.show()

plt.figure(figsize=(6,4))
sns.violinplot(data=df, x="PlacementStatus", y="AttendancePercent")
plt.title("Attendance vs Placement Status")
plt.show()

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Gender", hue="PlacementStatus")
plt.title("Gender vs Placement Status")
plt.show()
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="Stream", hue="PlacementStatus")
plt.xticks(rotation=45)
plt.title("Stream vs Placement Status")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include=["int64","float64"]).corr(),
            annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

sns.pairplot(df[["CGPA","AttendancePercent","CodingTestScore",
                 "AptitudeTestScore","PlacementStatus"]],
             hue="PlacementStatus")
plt.show()

print("\nEDA Completed Successfully")
```
